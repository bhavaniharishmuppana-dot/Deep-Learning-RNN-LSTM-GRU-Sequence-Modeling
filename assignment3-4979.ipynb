```python
# ============================================================
# PART A – RNN FOR SEQUENCE PREDICTION
# Q1, Q2, Q3 and Q4
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import time

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense, Input
from tensorflow.keras.optimizers import Adam

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

print("TensorFlow version:", tf.__version__)


# ============================================================
# 2. CREATE REALISTIC TEMPERATURE TIME-SERIES DATASET
# ============================================================

np.random.seed(42)

num_days = 1500

dates = pd.date_range(
    start="2020-01-01",
    periods=num_days,
    freq="D"
)

days = np.arange(num_days)

# Seasonal temperature variation
seasonal = 8 * np.sin(2 * np.pi * days / 365)

# Short-term temperature variation
short_term = 2 * np.sin(2 * np.pi * days / 30)

# Random variation
noise = np.random.normal(0, 1.5, num_days)

# Temperature values
temperature_values = (
    25 +
    seasonal +
    short_term +
    noise
)

df = pd.DataFrame({
    "Date": dates,
    "Temp": temperature_values
})

print("\nDataset created successfully!")
print("Dataset shape:", df.shape)

display(df.head())


# ============================================================
# 3. VISUALIZE DATA
# ============================================================

plt.figure(figsize=(14, 5))

plt.plot(
    df["Date"],
    df["Temp"]
)

plt.title("Daily Temperature Time Series")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.grid(True)

plt.show()


# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

print("\nMissing values:")
print(df.isnull().sum())

temperature = df["Temp"].values.astype("float32")

temperature = temperature.reshape(-1, 1)

print("\nTemperature statistics:")
print("Minimum:", temperature.min())
print("Maximum:", temperature.max())
print("Mean:", temperature.mean())


# ============================================================
# 5. NORMALIZATION
# ============================================================

scaler = MinMaxScaler(
    feature_range=(0, 1)
)

temperature_scaled = scaler.fit_transform(
    temperature
)

print("\nAfter normalization:")
print("Minimum:", temperature_scaled.min())
print("Maximum:", temperature_scaled.max())


# ============================================================
# 6. TRAIN-TEST SPLIT
# ============================================================

train_size = int(
    len(temperature_scaled) * 0.80
)

train_data = temperature_scaled[:train_size]

test_data = temperature_scaled[train_size:]

print("\nTrain-Test Split")
print("Total samples:", len(temperature_scaled))
print("Training samples:", len(train_data))
print("Testing samples:", len(test_data))


# ============================================================
# 7. SLIDING WINDOW SEQUENCE GENERATION
# ============================================================

def create_sequences(data, sequence_length):

    X = []
    y = []

    for i in range(
        len(data) - sequence_length
    ):

        X.append(
            data[i:i + sequence_length]
        )

        y.append(
            data[i + sequence_length]
        )

    return np.array(X), np.array(y)


sequence_length = 30

X_train, y_train = create_sequences(
    train_data,
    sequence_length
)

test_with_history = temperature_scaled[
    train_size - sequence_length:
]

X_test, y_test = create_sequences(
    test_with_history,
    sequence_length
)

print("\nSequence Generation")
print("Sequence length:", sequence_length)
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)


# ============================================================
# 8. BUILD SIMPLE RNN MODEL
# ============================================================

model = Sequential([

    Input(
        shape=(sequence_length, 1)
    ),

    SimpleRNN(
        units=64,
        activation="tanh"
    ),

    Dense(1)
])

model.compile(
    optimizer=Adam(
        learning_rate=0.001
    ),
    loss="mse"
)

print("\nSimple RNN Model:")
model.summary()


# ============================================================
# 9. TRAIN SIMPLE RNN
# ============================================================

start_time = time.time()

history = model.fit(

    X_train,
    y_train,

    epochs=30,

    batch_size=32,

    validation_split=0.1,

    verbose=1
)

training_time = (
    time.time() - start_time
)

print(
    "\nTraining time:",
    round(training_time, 2),
    "seconds"
)


# ============================================================
# 10. TRAINING AND VALIDATION LOSS
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "Training and Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")

plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# 11. MAKE PREDICTIONS
# ============================================================

predictions_scaled = model.predict(
    X_test,
    verbose=0
)

# Convert normalized values back
# to original temperature values

predictions = scaler.inverse_transform(
    predictions_scaled
)

actual = scaler.inverse_transform(
    y_test
)

print("\nPrediction completed!")


# ============================================================
# 12. Q3 – MODEL EVALUATION
# ============================================================

mae = mean_absolute_error(
    actual,
    predictions
)

mse = mean_squared_error(
    actual,
    predictions
)

rmse = np.sqrt(mse)

print("\n================================")
print("RNN MODEL PERFORMANCE")
print("================================")

print(
    f"MAE  : {mae:.4f} °C"
)

print(
    f"MSE  : {mse:.4f}"
)

print(
    f"RMSE : {rmse:.4f} °C"
)


# ============================================================
# 13. ACTUAL VS PREDICTED GRAPH
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    actual.flatten(),
    label="Actual Temperature"
)

plt.plot(
    predictions.flatten(),
    label="Predicted Temperature"
)

plt.title(
    "Actual vs Predicted Temperature"
)

plt.xlabel("Test Sample")
plt.ylabel("Temperature (°C)")

plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# 14. ACTUAL VS PREDICTED TABLE
# ============================================================

results = pd.DataFrame({

    "Actual Temperature (°C)":
        actual.flatten(),

    "Predicted Temperature (°C)":
        predictions.flatten()

})

results["Absolute Error (°C)"] = abs(
    results["Actual Temperature (°C)"]
    -
    results["Predicted Temperature (°C)"]
)

print("\nFirst 20 Predictions:")

display(
    results.head(20).round(3)
)


# ============================================================
# 15. Q2 – HYPERPARAMETER EXPERIMENT FUNCTION
# ============================================================

def train_experiment(
    sequence_length=30,
    hidden_units=64,
    learning_rate=0.001,
    batch_size=32,
    epochs=10
):

    # Create sequences

    X_train_exp, y_train_exp = (
        create_sequences(
            train_data,
            sequence_length
        )
    )

    test_history_exp = (
        temperature_scaled[
            train_size - sequence_length:
        ]
    )

    X_test_exp, y_test_exp = (
        create_sequences(
            test_history_exp,
            sequence_length
        )
    )

    # Build model

    exp_model = Sequential([

        Input(
            shape=(sequence_length, 1)
        ),

        SimpleRNN(
            units=hidden_units,
            activation="tanh"
        ),

        Dense(1)
    ])

    # Compile

    exp_model.compile(

        optimizer=Adam(
            learning_rate=learning_rate
        ),

        loss="mse"
    )

    # Train

    start = time.time()

    exp_model.fit(

        X_train_exp,
        y_train_exp,

        epochs=epochs,

        batch_size=batch_size,

        validation_split=0.1,

        verbose=0
    )

    training_time_exp = (
        time.time() - start
    )

    # Predict

    pred_scaled = exp_model.predict(
        X_test_exp,
        verbose=0
    )

    pred = scaler.inverse_transform(
        pred_scaled
    )

    actual_exp = scaler.inverse_transform(
        y_test_exp
    )

    # Metrics

    mae_exp = mean_absolute_error(
        actual_exp,
        pred
    )

    mse_exp = mean_squared_error(
        actual_exp,
        pred
    )

    rmse_exp = np.sqrt(
        mse_exp
    )

    return {

        "Sequence Length":
            sequence_length,

        "Hidden Units":
            hidden_units,

        "Learning Rate":
            learning_rate,

        "Batch Size":
            batch_size,

        "Epochs":
            epochs,

        "MAE":
            mae_exp,

        "MSE":
            mse_exp,

        "RMSE":
            rmse_exp,

        "Training Time":
            training_time_exp
    }


# ============================================================
# 16. RUN HYPERPARAMETER EXPERIMENTS
# ============================================================

experiments = []


# Sequence Length
for seq in [10, 30, 60]:

    print(
        "Testing sequence length:",
        seq
    )

    result = train_experiment(

        sequence_length=seq,

        hidden_units=64,

        learning_rate=0.001,

        batch_size=32,

        epochs=10
    )

    experiments.append(result)


# Hidden Units
for units in [32, 64, 128]:

    print(
        "Testing hidden units:",
        units
    )

    result = train_experiment(

        sequence_length=30,

        hidden_units=units,

        learning_rate=0.001,

        batch_size=32,

        epochs=10
    )

    experiments.append(result)


# Learning Rate
for lr in [0.0001, 0.001, 0.01]:

    print(
        "Testing learning rate:",
        lr
    )

    result = train_experiment(

        sequence_length=30,

        hidden_units=64,

        learning_rate=lr,

        batch_size=32,

        epochs=10
    )

    experiments.append(result)


# Batch Size
for batch in [16, 32, 64]:

    print(
        "Testing batch size:",
        batch
    )

    result = train_experiment(

        sequence_length=30,

        hidden_units=64,

        learning_rate=0.001,

        batch_size=batch,

        epochs=10
    )

    experiments.append(result)


# Epochs
for epoch in [5, 10, 20]:

    print(
        "Testing epochs:",
        epoch
    )

    result = train_experiment(

        sequence_length=30,

        hidden_units=64,

        learning_rate=0.001,

        batch_size=32,

        epochs=epoch
    )

    experiments.append(result)


# ============================================================
# 17. DISPLAY HYPERPARAMETER RESULTS
# ============================================================

experiment_results = pd.DataFrame(
    experiments
)

print(
    "\n=========================================="
)

print(
    "HYPERPARAMETER EXPERIMENT RESULTS"
)

print(
    "=========================================="
)

display(
    experiment_results.round(4)
)


# ============================================================
# 18. Q2 – RMSE COMPARISON GRAPH
# ============================================================

plt.figure(figsize=(14, 6))

plt.bar(
    range(
        len(experiment_results)
    ),

    experiment_results["RMSE"]
)

plt.xlabel(
    "Experiment Number"
)

plt.ylabel(
    "RMSE"
)

plt.title(
    "Effect of Hyperparameters on RMSE"
)

plt.xticks(
    range(
        len(experiment_results)
    ),

    [
        f"E{i+1}"
        for i in range(
            len(experiment_results)
        )
    ]
)

plt.grid(
    axis="y"
)

plt.show()


# ============================================================
# 19. BEST EXPERIMENT
# ============================================================

best_result = experiment_results.loc[
    experiment_results["RMSE"].idxmin()
]

print(
    "\nBest experiment based on RMSE:"
)

display(
    best_result.to_frame().T.round(4)
)


# ============================================================
# 20. Q4 – HIDDEN STATE DEMONSTRATION
# ============================================================

# Create a model that returns hidden states
# for every time step.

hidden_state_model = Sequential([

    Input(
        shape=(sequence_length, 1)
    ),

    SimpleRNN(
        units=8,
        activation="tanh",
        return_sequences=True
    )
])

# Select one test sequence

sample_sequence = X_test[0:1]

# Generate hidden states

hidden_states = (
    hidden_state_model.predict(
        sample_sequence,
        verbose=0
    )
)

print(
    "\nHidden state shape:"
)

print(
    hidden_states.shape
)

print(
    "\nThe hidden state is updated at every "
    "time step and carries information from "
    "previous time steps."
)


# ============================================================
# 21. VISUALIZE HIDDEN STATES
# ============================================================

plt.figure(figsize=(14, 6))

plt.plot(
    hidden_states[0]
)

plt.title(
    "Hidden State Values Across Time Steps"
)

plt.xlabel(
    "Time Step"
)

plt.ylabel(
    "Hidden State Value"
)

plt.grid(True)

plt.show()


# ============================================================
# 22. DISPLAY Q4 EXPLANATION
# ============================================================

print("\n==========================================")
print("Q4 – HIDDEN STATE AND VANISHING GRADIENT")
print("==========================================")

print("""
The Simple RNN maintains a hidden state that is
updated at every time step. The hidden state contains
information from the current input as well as
information carried from previous time steps.

Therefore, the RNN can use previous temperature
observations when predicting the next temperature.

However, during Backpropagation Through Time (BPTT),
gradients are repeatedly multiplied through the
recurrent connections. If these gradients become very
small, the vanishing-gradient problem occurs.

Because of this problem, a Simple RNN may have
difficulty learning long-term dependencies. Information
from earlier time steps can gradually lose its influence
on the current prediction.

LSTM and GRU architectures address this problem using
specialized gating mechanisms that help preserve useful
information for longer periods.
""")


# ============================================================
# 23. FINAL SUMMARY
# ============================================================

print("\n==========================================")
print("PART A – FINAL SUMMARY")
print("==========================================")

print("Application: Daily Temperature Prediction")
print("Architecture: Simple RNN")
print("Sequence Length:", sequence_length)
print("Hidden Units: 64")
print("Learning Rate: 0.001")
print("Batch Size: 32")
print("Epochs: 30")

print("\nFinal Model Metrics:")
print(f"MAE  = {mae:.4f} °C")
print(f"MSE  = {mse:.4f}")
print(f"RMSE = {rmse:.4f} °C")

print("\nPart A Q1-Q4 completed successfully!")
```

```python
# ============================================================
# PART B – WORD EMBEDDINGS AND RNN
# Q5, Q6, Q7 and Q8
# NO INTERNET REQUIRED
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import time
import re

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)

print("TensorFlow version:", tf.__version__)


# ============================================================
# Q5 – CREATE TEXT DATASET
# ============================================================

positive_reviews = [
    "This movie was excellent and amazing",
    "I really enjoyed this wonderful movie",
    "The film was fantastic and very entertaining",
    "Amazing acting and beautiful story",
    "I loved this movie very much",
    "The movie was excellent and enjoyable",
    "Fantastic movie with great acting",
    "The story was beautiful and touching",
    "This was a wonderful film",
    "The movie was brilliant and exciting",
    "I enjoyed every moment of this film",
    "Excellent performance by all actors",
    "Amazing story and great direction",
    "This movie is very good",
    "A fantastic and enjoyable experience",
    "The film was wonderful",
    "Great movie with excellent acting",
    "I loved the beautiful story",
    "Very entertaining and exciting movie",
    "Brilliant acting and fantastic direction",
    "This film was really enjoyable",
    "Amazing movie with a wonderful story",
    "Excellent film and great performance",
    "I liked this movie a lot",
    "The movie was interesting and exciting",
    "Wonderful acting and beautiful scenes",
    "Fantastic story and excellent music",
    "This is a great movie",
    "I really loved the film",
    "The movie was very entertaining"
]

negative_reviews = [
    "This movie was terrible and boring",
    "I really disliked this horrible movie",
    "The film was disappointing and boring",
    "Bad acting and terrible story",
    "I hated this movie",
    "The movie was poor and disappointing",
    "Terrible movie with bad acting",
    "The story was boring and weak",
    "This was a horrible film",
    "The movie was awful and boring",
    "I did not enjoy this film",
    "Very poor performance by the actors",
    "Bad story and terrible direction",
    "This movie is very bad",
    "A boring and disappointing experience",
    "The film was terrible",
    "Bad movie with poor acting",
    "I disliked the weak story",
    "Very boring and disappointing movie",
    "Awful acting and terrible direction",
    "This film was really boring",
    "Terrible movie with a weak story",
    "Poor film and bad performance",
    "I hated this movie a lot",
    "The movie was boring and disappointing",
    "Horrible acting and poor scenes",
    "Bad story and terrible music",
    "This is a bad movie",
    "I really disliked the film",
    "The movie was not entertaining"
]


# Combine dataset
texts = positive_reviews + negative_reviews
labels = [1] * len(positive_reviews) + [0] * len(negative_reviews)

data = pd.DataFrame({
    "Review": texts,
    "Label": labels
})

print("\nDataset created successfully!")
print("Total reviews:", len(data))

display(data.head(10))


# ============================================================
# TEXT CLEANING
# ============================================================

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()


data["Cleaned Review"] = data["Review"].apply(clean_text)

print("\nAfter text cleaning:")
display(data.head(10))


# ============================================================
# TOKENIZATION / INTEGER ENCODING
# ============================================================

vocab_size = 1000
max_length = 15

tokenizer = Tokenizer(
    num_words=vocab_size,
    oov_token="<OOV>"
)

tokenizer.fit_on_texts(data["Cleaned Review"])

sequences = tokenizer.texts_to_sequences(
    data["Cleaned Review"]
)

print("\nInteger encoded example:")
print(sequences[0])


# ============================================================
# PADDING
# ============================================================

X = pad_sequences(
    sequences,
    maxlen=max_length,
    padding="post",
    truncating="post"
)

y = np.array(labels)

print("\nPadded data shape:")
print("X:", X.shape)
print("y:", y.shape)

print("\nFirst padded sequence:")
print(X[0])


# ============================================================
# TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("\nTrain-Test Split")
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))


# ============================================================
# Q5 – SIMPLE RNN TEXT CLASSIFICATION
# ============================================================

print("\n==========================================")
print("Q5 – SIMPLE RNN TEXT CLASSIFICATION")
print("==========================================")

embedding_dim = 64

rnn_model = Sequential([
    Input(shape=(max_length,)),

    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim,
        trainable=True
    ),

    SimpleRNN(
        32,
        activation="tanh"
    ),

    Dense(
        1,
        activation="sigmoid"
    )
])

rnn_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

rnn_model.summary()


# ============================================================
# TRAIN RNN
# ============================================================

start_time = time.time()

rnn_history = rnn_model.fit(
    X_train,
    y_train,
    epochs=20,
    batch_size=8,
    validation_split=0.2,
    verbose=1
)

rnn_training_time = time.time() - start_time

print("\nTraining time:",
      round(rnn_training_time, 2),
      "seconds")


# ============================================================
# TRAINING / VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    rnn_history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    rnn_history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("RNN Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# TRAINING / VALIDATION LOSS
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    rnn_history.history["loss"],
    label="Training Loss"
)

plt.plot(
    rnn_history.history["val_loss"],
    label="Validation Loss"
)

plt.title("RNN Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# RNN PREDICTIONS
# ============================================================

probabilities = rnn_model.predict(
    X_test,
    verbose=0
)

predictions = (
    probabilities >= 0.5
).astype(int).flatten()


# ============================================================
# Q7 – EVALUATION METRICS
# ============================================================

accuracy = accuracy_score(
    y_test,
    predictions
)

precision = precision_score(
    y_test,
    predictions,
    zero_division=0
)

recall = recall_score(
    y_test,
    predictions,
    zero_division=0
)

f1 = f1_score(
    y_test,
    predictions,
    zero_division=0
)

print("\n==========================================")
print("Q7 – RNN PERFORMANCE")
print("==========================================")

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")


# ============================================================
# CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    predictions
)

print("\nConfusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Negative", "Positive"]
)

disp.plot(values_format="d")

plt.title("RNN Confusion Matrix")
plt.show()


# ============================================================
# Q6 – THREE EMBEDDING DIMENSIONS
# ============================================================

print("\n==========================================")
print("Q6 – EMBEDDING DIMENSION EXPERIMENT")
print("==========================================")

embedding_dimensions = [16, 32, 64]

embedding_results = []

for dimension in embedding_dimensions:

    print("\n--------------------------------------")
    print("Embedding Dimension:", dimension)
    print("--------------------------------------")

    model = Sequential([
        Input(shape=(max_length,)),

        Embedding(
            input_dim=vocab_size,
            output_dim=dimension,
            trainable=True
        ),

        SimpleRNN(
            32,
            activation="tanh"
        ),

        Dense(
            1,
            activation="sigmoid"
        )
    ])

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    start_time = time.time()

    history = model.fit(
        X_train,
        y_train,
        epochs=15,
        batch_size=8,
        validation_split=0.2,
        verbose=0
    )

    training_time = time.time() - start_time

    probabilities = model.predict(
        X_test,
        verbose=0
    )

    predictions_exp = (
        probabilities >= 0.5
    ).astype(int).flatten()

    accuracy_exp = accuracy_score(
        y_test,
        predictions_exp
    )

    precision_exp = precision_score(
        y_test,
        predictions_exp,
        zero_division=0
    )

    recall_exp = recall_score(
        y_test,
        predictions_exp,
        zero_division=0
    )

    f1_exp = f1_score(
        y_test,
        predictions_exp,
        zero_division=0
    )

    embedding_results.append({
        "Embedding Dimension": dimension,
        "Accuracy": accuracy_exp,
        "Precision": precision_exp,
        "Recall": recall_exp,
        "F1 Score": f1_exp,
        "Training Time (sec)": training_time
    })


# ============================================================
# EMBEDDING RESULTS
# ============================================================

embedding_results_df = pd.DataFrame(
    embedding_results
)

print("\n==========================================")
print("EMBEDDING DIMENSION RESULTS")
print("==========================================")

display(
    embedding_results_df.round(4)
)


# ============================================================
# EMBEDDING DIMENSION vs ACCURACY
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    embedding_results_df["Embedding Dimension"],
    embedding_results_df["Accuracy"],
    marker="o"
)

plt.title("Embedding Dimension vs Accuracy")
plt.xlabel("Embedding Dimension")
plt.ylabel("Accuracy")
plt.grid(True)
plt.show()


# ============================================================
# EMBEDDING DIMENSION vs F1 SCORE
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    embedding_results_df["Embedding Dimension"],
    embedding_results_df["F1 Score"],
    marker="o"
)

plt.title("Embedding Dimension vs F1 Score")
plt.xlabel("Embedding Dimension")
plt.ylabel("F1 Score")
plt.grid(True)
plt.show()


# ============================================================
# Q8 – WORD EMBEDDINGS VS ONE-HOT ENCODING
# ============================================================

print("\n==========================================")
print("Q8 – WORD EMBEDDINGS VS ONE-HOT ENCODING")
print("==========================================")

print("""
WORD EMBEDDINGS:
1. Words are represented using dense numerical vectors.
2. The vectors are learned during model training.
3. Similar words can have similar vector representations.
4. Embeddings require less memory than large one-hot vectors.
5. Embeddings can capture useful semantic relationships.

ONE-HOT ENCODING:
1. Each word is represented by a binary vector.
2. Only one position contains 1.
3. The remaining positions contain 0.
4. The vectors become very large for a large vocabulary.
5. One-hot encoding does not directly represent semantic similarity.

Therefore, trainable word embeddings are more suitable for
RNN-based text classification because they provide compact,
learnable representations of words.
""")


# ============================================================
# FINAL PART B SUMMARY
# ============================================================

print("\n==========================================")
print("PART B – FINAL SUMMARY")
print("==========================================")

print("Dataset       : Movie Review Sentiment Dataset")
print("Task          : Sentiment Classification")
print("Vocabulary    :", vocab_size)
print("Sequence Size :", max_length)
print("Embedding     :", embedding_dim)
print("RNN Units     : 32")
print("Epochs        : 20")

print("\nBase RNN Performance:")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")

print("\nEmbedding dimensions tested:",
      embedding_dimensions)

print("\nPart B Q5-Q8 completed successfully!")
```

```python
# ============================================================
# PART C – LSTM AND GRU
# Q9, Q10, Q11, Q12, Q13 and Q14
# NO INTERNET REQUIRED
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import time
import re

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Embedding,
    SimpleRNN,
    LSTM,
    GRU,
    Dense
)
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("TensorFlow version:", tf.__version__)


# ============================================================
# CREATE DATASET
# ============================================================

positive_reviews = [
    "This movie was excellent and amazing",
    "I really enjoyed this wonderful movie",
    "The film was fantastic and very entertaining",
    "Amazing acting and beautiful story",
    "I loved this movie very much",
    "The movie was excellent and enjoyable",
    "Fantastic movie with great acting",
    "The story was beautiful and touching",
    "This was a wonderful film",
    "The movie was brilliant and exciting",
    "I enjoyed every moment of this film",
    "Excellent performance by all actors",
    "Amazing story and great direction",
    "This movie is very good",
    "A fantastic and enjoyable experience",
    "The film was wonderful",
    "Great movie with excellent acting",
    "I loved the beautiful story",
    "Very entertaining and exciting movie",
    "Brilliant acting and fantastic direction",
    "This film was really enjoyable",
    "Amazing movie with a wonderful story",
    "Excellent film and great performance",
    "I liked this movie a lot",
    "The movie was interesting and exciting",
    "Wonderful acting and beautiful scenes",
    "Fantastic story and excellent music",
    "This is a great movie",
    "I really loved the film",
    "The movie was very entertaining"
]

negative_reviews = [
    "This movie was terrible and boring",
    "I really disliked this horrible movie",
    "The film was disappointing and boring",
    "Bad acting and terrible story",
    "I hated this movie",
    "The movie was poor and disappointing",
    "Terrible movie with bad acting",
    "The story was boring and weak",
    "This was a horrible film",
    "The movie was awful and boring",
    "I did not enjoy this film",
    "Very poor performance by the actors",
    "Bad story and terrible direction",
    "This movie is very bad",
    "A boring and disappointing experience",
    "The film was terrible",
    "Bad movie with poor acting",
    "I disliked the weak story",
    "Very boring and disappointing movie",
    "Awful acting and terrible direction",
    "This film was really boring",
    "Terrible movie with a weak story",
    "Poor film and bad performance",
    "I hated this movie a lot",
    "The movie was boring and disappointing",
    "Horrible acting and poor scenes",
    "Bad story and terrible music",
    "This is a bad movie",
    "I really disliked the film",
    "The movie was not entertaining"
]

texts = positive_reviews + negative_reviews
labels = [1] * len(positive_reviews) + [0] * len(negative_reviews)

data = pd.DataFrame({
    "Review": texts,
    "Label": labels
})


# ============================================================
# TEXT CLEANING
# ============================================================

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z\s]", "", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()


data["Cleaned Review"] = data["Review"].apply(clean_text)


# ============================================================
# TOKENIZATION
# ============================================================

vocab_size = 1000
max_length = 15

tokenizer = Tokenizer(
    num_words=vocab_size,
    oov_token="<OOV>"
)

tokenizer.fit_on_texts(
    data["Cleaned Review"]
)

sequences = tokenizer.texts_to_sequences(
    data["Cleaned Review"]
)

X = pad_sequences(
    sequences,
    maxlen=max_length,
    padding="post",
    truncating="post"
)

y = np.array(labels)


# ============================================================
# TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("\nDataset prepared successfully!")
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Vocabulary size:", vocab_size)
print("Sequence length:", max_length)


# ============================================================
# FUNCTION TO TRAIN AND EVALUATE MODEL
# ============================================================

def train_model(model, model_name, epochs=20):

    print("\n==========================================")
    print(model_name)
    print("==========================================")

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    start_time = time.time()

    history = model.fit(
        X_train,
        y_train,
        epochs=epochs,
        batch_size=8,
        validation_split=0.2,
        verbose=0
    )

    training_time = time.time() - start_time

    probabilities = model.predict(
        X_test,
        verbose=0
    )

    predictions = (
        probabilities >= 0.5
    ).astype(int).flatten()

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    loss, test_accuracy = model.evaluate(
        X_test,
        y_test,
        verbose=0
    )

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))
    print("Test Loss:", round(loss, 4))
    print("Parameters:", model.count_params())
    print("Training Time:", round(training_time, 2), "seconds")

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "Test Loss": loss,
        "Parameters": model.count_params(),
        "Training Time (sec)": training_time,
        "History": history,
        "Model Object": model
    }


# ============================================================
# Q9 – SIMPLE RNN
# ============================================================

rnn_model = Sequential([
    Input(shape=(max_length,)),

    Embedding(
        input_dim=vocab_size,
        output_dim=64,
        trainable=True
    ),

    SimpleRNN(
        32,
        activation="tanh"
    ),

    Dense(
        1,
        activation="sigmoid"
    )
])

rnn_result = train_model(
    rnn_model,
    "Simple RNN"
)


# ============================================================
# Q9 – LSTM
# ============================================================

lstm_model = Sequential([
    Input(shape=(max_length,)),

    Embedding(
        input_dim=vocab_size,
        output_dim=64,
        trainable=True
    ),

    LSTM(
        32
    ),

    Dense(
        1,
        activation="sigmoid"
    )
])

lstm_result = train_model(
    lstm_model,
    "LSTM"
)


# ============================================================
# Q12 – GRU
# ============================================================

gru_model = Sequential([
    Input(shape=(max_length,)),

    Embedding(
        input_dim=vocab_size,
        output_dim=64,
        trainable=True
    ),

    GRU(
        32
    ),

    Dense(
        1,
        activation="sigmoid"
    )
])

gru_result = train_model(
    gru_model,
    "GRU"
)


# ============================================================
# COMPARISON TABLE
# ============================================================

results = [
    rnn_result,
    lstm_result,
    gru_result
]

comparison_df = pd.DataFrame([
    {
        "Model": r["Model"],
        "Accuracy": r["Accuracy"],
        "Precision": r["Precision"],
        "Recall": r["Recall"],
        "F1 Score": r["F1 Score"],
        "Test Loss": r["Test Loss"],
        "Parameters": r["Parameters"],
        "Training Time (sec)": r["Training Time (sec)"]
    }
    for r in results
])

print("\n==========================================")
print("Q13 – RNN vs LSTM vs GRU")
print("==========================================")

display(
    comparison_df.round(4)
)


# ============================================================
# ACCURACY COMPARISON
# ============================================================

plt.figure(figsize=(9, 5))

plt.bar(
    comparison_df["Model"],
    comparison_df["Accuracy"]
)

plt.title("RNN vs LSTM vs GRU – Accuracy")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.grid(axis="y")

plt.show()


# ============================================================
# LOSS COMPARISON
# ============================================================

plt.figure(figsize=(9, 5))

plt.bar(
    comparison_df["Model"],
    comparison_df["Test Loss"]
)

plt.title("RNN vs LSTM vs GRU – Test Loss")
plt.xlabel("Model")
plt.ylabel("Loss")
plt.grid(axis="y")

plt.show()


# ============================================================
# PARAMETER COMPARISON
# ============================================================

plt.figure(figsize=(9, 5))

plt.bar(
    comparison_df["Model"],
    comparison_df["Parameters"]
)

plt.title("RNN vs LSTM vs GRU – Number of Parameters")
plt.xlabel("Model")
plt.ylabel("Parameters")
plt.grid(axis="y")

plt.show()


# ============================================================
# TRAINING TIME COMPARISON
# ============================================================

plt.figure(figsize=(9, 5))

plt.bar(
    comparison_df["Model"],
    comparison_df["Training Time (sec)"]
)

plt.title("RNN vs LSTM vs GRU – Training Time")
plt.xlabel("Model")
plt.ylabel("Time (seconds)")
plt.grid(axis="y")

plt.show()


# ============================================================
# LOSS CURVES
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    rnn_result["History"].history["val_loss"],
    label="Simple RNN"
)

plt.plot(
    lstm_result["History"].history["val_loss"],
    label="LSTM"
)

plt.plot(
    gru_result["History"].history["val_loss"],
    label="GRU"
)

plt.title("Validation Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# ACCURACY CURVES
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    rnn_result["History"].history["val_accuracy"],
    label="Simple RNN"
)

plt.plot(
    lstm_result["History"].history["val_accuracy"],
    label="LSTM"
)

plt.plot(
    gru_result["History"].history["val_accuracy"],
    label="GRU"
)

plt.title("Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# Q10 – LSTM HIDDEN STATE AND CELL STATE
# ============================================================

print("\n==========================================")
print("Q10 – LSTM HIDDEN STATE AND CELL STATE")
print("==========================================")

lstm_state_model = Sequential([
    Input(shape=(max_length,)),

    Embedding(
        input_dim=vocab_size,
        output_dim=64
    ),

    LSTM(
        32,
        return_state=True
    )
])

lstm_output, hidden_state, cell_state = lstm_state_model.predict(
    X_test[:1],
    verbose=0
)

print("\nLSTM Output Shape:")
print(lstm_output.shape)

print("\nHidden State Shape:")
print(hidden_state.shape)

print("\nCell State Shape:")
print(cell_state.shape)

print("""
The LSTM maintains two important states:

1. Hidden State (h):
   Contains the current output information of the LSTM.

2. Cell State (c):
   Acts as a long-term memory pathway and carries useful
   information across time steps.

The LSTM uses gates to control which information is forgotten,
stored, and passed to the next time step.
""")


# ============================================================
# Q11 – LSTM GATES
# ============================================================

print("\n==========================================")
print("Q11 – LSTM GATES")
print("==========================================")

print("""
LSTM uses three main gates:

1. FORGET GATE:
   Decides which information from the previous cell state
   should be removed.

2. INPUT GATE:
   Decides which new information should be stored in the
   cell state.

3. OUTPUT GATE:
   Decides which information from the cell state should
   be used to produce the current hidden state.

The cell state provides a pathway for carrying information
over long sequences, while the gates control the flow of
information.

This mechanism helps LSTM handle long-term dependencies
better than a basic Simple RNN.
""")


# ============================================================
# Q12 – GRU
# ============================================================

print("\n==========================================")
print("Q12 – GRU")
print("==========================================")

print("""
GRU (Gated Recurrent Unit) is another recurrent neural
network architecture designed to handle long-term
dependencies.

GRU uses two main gates:

1. UPDATE GATE:
   Controls how much previous information should be retained
   and how much new information should be added.

2. RESET GATE:
   Controls how much previous information should be forgotten
   when processing new input.

Unlike LSTM, GRU does not maintain a separate cell state.
It uses a hidden state to store information.

GRU generally has fewer parameters than LSTM, which can make
it computationally efficient while still handling long-term
dependencies.
""")


# ============================================================
# Q13 – LONG-TERM DEPENDENCY EXPLANATION
# ============================================================

print("\n==========================================")
print("Q13 – LONG-TERM DEPENDENCY COMPARISON")
print("==========================================")

print("""
Simple RNN:
• Simple architecture
• Fewer parameters
• Can suffer from vanishing gradients
• May have difficulty learning long-term dependencies

LSTM:
• Uses forget, input and output gates
• Maintains hidden state and cell state
• Better suited for long-term dependencies
• More parameters than Simple RNN

GRU:
• Uses update and reset gates
• Maintains a hidden state
• Fewer parameters than LSTM
• Can handle long-term dependencies efficiently
""")


# ============================================================
# Q14 – ARCHITECTURE SELECTION
# ============================================================

print("\n==========================================")
print("Q14 – ARCHITECTURE ANALYSIS")
print("==========================================")

best_accuracy_model = comparison_df.loc[
    comparison_df["Accuracy"].idxmax(),
    "Model"
]

lowest_loss_model = comparison_df.loc[
    comparison_df["Test Loss"].idxmin(),
    "Model"
]

print("Model with highest test accuracy:",
      best_accuracy_model)

print("Model with lowest test loss:",
      lowest_loss_model)

print("""
Architecture selection should consider accuracy, loss,
training time, number of parameters, and the ability to
handle long-term dependencies.

Simple RNN is computationally simpler but can suffer from
vanishing gradients.

LSTM is useful when learning long-term dependencies is
important because it uses a cell state and three gates.

GRU provides a simpler gated architecture with fewer
parameters than LSTM while still handling long-term
dependencies.

The experimental comparison above should be used to justify
the architecture selected for this particular dataset.
""")


# ============================================================
# FINAL PART C SUMMARY
# ============================================================

print("\n==========================================")
print("PART C – FINAL SUMMARY")
print("==========================================")

print("Dataset       : Movie Review Sentiment Dataset")
print("Task          : Sentiment Classification")
print("Vocabulary    :", vocab_size)
print("Sequence Size :", max_length)
print("Embedding     : 64")
print("Hidden Units  : 32")
print("Epochs        : 20")

print("\nModels compared:")
print("1. Simple RNN")
print("2. LSTM")
print("3. GRU")

print("\nComparison Results:")
display(comparison_df.round(4))

print("\nPart C Q9-Q14 completed successfully!")
```

```python
# ============================================================
# PART D – ENCODER-DECODER SEQUENCE TO SEQUENCE TRANSLATION
# Q15, Q16, Q17, Q18, Q19 and Q20
# ENGLISH -> GERMAN
# NO INTERNET REQUIRED
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import re
import time

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input,
    LSTM,
    Embedding,
    Dense
)

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

print("TensorFlow version:", tf.__version__)


# ============================================================
# Q15 – PARALLEL ENGLISH-GERMAN DATASET
# ============================================================

english_sentences = [
    "hello",
    "good morning",
    "good night",
    "thank you",
    "how are you",
    "i am fine",
    "what is your name",
    "my name is harish",
    "where are you",
    "i am at home",
    "i like music",
    "i like movies",
    "i like cricket",
    "i love my family",
    "this is good",
    "this is bad",
    "the weather is good",
    "the weather is cold",
    "i am happy",
    "i am sad",
    "i am hungry",
    "i am tired",
    "please help me",
    "come here",
    "go home",
    "see you tomorrow",
    "see you soon",
    "have a nice day",
    "what are you doing",
    "i am studying",
    "i am learning",
    "i am working",
    "this is my book",
    "this is my house",
    "where is the school",
    "where is the hospital",
    "i need water",
    "i need food",
    "open the door",
    "close the door"
]

german_sentences = [
    "hallo",
    "guten morgen",
    "gute nacht",
    "danke",
    "wie geht es dir",
    "mir geht es gut",
    "wie heißt du",
    "mein name ist harish",
    "wo bist du",
    "ich bin zu hause",
    "ich mag musik",
    "ich mag filme",
    "ich mag cricket",
    "ich liebe meine familie",
    "das ist gut",
    "das ist schlecht",
    "das wetter ist gut",
    "das wetter ist kalt",
    "ich bin glücklich",
    "ich bin traurig",
    "ich habe hunger",
    "ich bin müde",
    "bitte hilf mir",
    "komm hierher",
    "geh nach hause",
    "bis morgen",
    "bis bald",
    "einen schönen tag",
    "was machst du",
    "ich lerne",
    "ich lerne gerade",
    "ich arbeite",
    "das ist mein buch",
    "das ist mein haus",
    "wo ist die schule",
    "wo ist das krankenhaus",
    "ich brauche wasser",
    "ich brauche essen",
    "öffne die tür",
    "schließe die tür"
]

data = pd.DataFrame({
    "English": english_sentences,
    "German": german_sentences
})

print("\nParallel dataset created!")
print("Number of sentence pairs:", len(data))

display(data.head(10))


# ============================================================
# TEXT PREPROCESSING
# ============================================================

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-ZäöüÄÖÜß\s]", "", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()


data["English"] = data["English"].apply(clean_text)
data["German"] = data["German"].apply(clean_text)


# ============================================================
# Q18 – START AND END TOKENS
# ============================================================

data["German_Input"] = (
    "<start> " + data["German"] + " <end>"
)

data["German_Target"] = (
    data["German"] + " <end>"
)

print("\nExample with START and END tokens:")
display(
    data[["English", "German_Input", "German_Target"]].head(5)
)


# ============================================================
# TOKENIZATION
# ============================================================

english_tokenizer = Tokenizer(
    filters=""
)

german_tokenizer = Tokenizer(
    filters=""
)

english_tokenizer.fit_on_texts(
    data["English"]
)

german_tokenizer.fit_on_texts(
    data["German_Input"]
)

encoder_sequences = english_tokenizer.texts_to_sequences(
    data["English"]
)

decoder_input_sequences = german_tokenizer.texts_to_sequences(
    data["German_Input"]
)

decoder_target_sequences = german_tokenizer.texts_to_sequences(
    data["German_Target"]
)


# ============================================================
# PADDING
# ============================================================

max_encoder_length = max(
    len(x) for x in encoder_sequences
)

max_decoder_length = max(
    len(x) for x in decoder_input_sequences
)

encoder_input_data = pad_sequences(
    encoder_sequences,
    maxlen=max_encoder_length,
    padding="post"
)

decoder_input_data = pad_sequences(
    decoder_input_sequences,
    maxlen=max_decoder_length,
    padding="post"
)

decoder_target_data = pad_sequences(
    decoder_target_sequences,
    maxlen=max_decoder_length,
    padding="post"
)


# Add last dimension for sparse categorical cross entropy
decoder_target_data = np.expand_dims(
    decoder_target_data,
    -1
)


encoder_vocab_size = len(
    english_tokenizer.word_index
) + 1

decoder_vocab_size = len(
    german_tokenizer.word_index
) + 1


print("\nTokenization completed!")

print("English vocabulary size:",
      encoder_vocab_size)

print("German vocabulary size:",
      decoder_vocab_size)

print("Encoder sequence length:",
      max_encoder_length)

print("Decoder sequence length:",
      max_decoder_length)


# ============================================================
# TRAIN-TEST SPLIT
# ============================================================

# Use most sentences for training
# Keep 10 sentences for unseen testing

train_size = len(data) - 10

encoder_train = encoder_input_data[:train_size]
decoder_train = decoder_input_data[:train_size]
target_train = decoder_target_data[:train_size]

encoder_test = encoder_input_data[train_size:]
decoder_test = decoder_input_data[train_size:]
target_test = decoder_target_data[train_size:]

english_test = data["English"].iloc[train_size:].tolist()
german_test = data["German"].iloc[train_size:].tolist()

print("\nTraining sentence pairs:", train_size)
print("Unseen test sentences:", len(english_test))


# ============================================================
# Q16 – ENCODER
# ============================================================

latent_dim = 64

encoder_inputs = Input(
    shape=(max_encoder_length,),
    name="Encoder_Input"
)

encoder_embedding = Embedding(
    input_dim=encoder_vocab_size,
    output_dim=64,
    name="Encoder_Embedding"
)(encoder_inputs)

encoder_lstm = LSTM(
    latent_dim,
    return_state=True,
    name="Encoder_LSTM"
)

encoder_outputs, state_h, state_c = encoder_lstm(
    encoder_embedding
)

encoder_states = [
    state_h,
    state_c
]


# ============================================================
# Q17 – DECODER
# ============================================================

decoder_inputs = Input(
    shape=(max_decoder_length,),
    name="Decoder_Input"
)

decoder_embedding_layer = Embedding(
    input_dim=decoder_vocab_size,
    output_dim=64,
    name="Decoder_Embedding"
)

decoder_embedding = decoder_embedding_layer(
    decoder_inputs
)

decoder_lstm = LSTM(
    latent_dim,
    return_sequences=True,
    return_state=True,
    name="Decoder_LSTM"
)

decoder_outputs, _, _ = decoder_lstm(
    decoder_embedding,
    initial_state=encoder_states
)

decoder_dense = Dense(
    decoder_vocab_size,
    activation="softmax",
    name="Output_Layer"
)

decoder_outputs = decoder_dense(
    decoder_outputs
)


# ============================================================
# COMPLETE ENCODER-DECODER MODEL
# ============================================================

seq2seq_model = Model(
    [encoder_inputs, decoder_inputs],
    decoder_outputs
)

seq2seq_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("\n==========================================")
print("ENCODER-DECODER MODEL")
print("==========================================")

seq2seq_model.summary()


# ============================================================
# TRAIN MODEL
# ============================================================

print("\nTraining Encoder-Decoder model...")

start_time = time.time()

history = seq2seq_model.fit(
    [encoder_train, decoder_train],
    target_train,
    batch_size=8,
    epochs=150,
    validation_split=0.15,
    verbose=0
)

training_time = time.time() - start_time

print("\nTraining completed!")
print("Training time:",
      round(training_time, 2),
      "seconds")


# ============================================================
# TRAINING LOSS
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Encoder-Decoder Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# Q16 – DISPLAY ENCODER STATES
# ============================================================

print("\n==========================================")
print("Q16 – ENCODER HIDDEN AND CELL STATES")
print("==========================================")

encoder_state_model = Model(
    encoder_inputs,
    encoder_states
)

sample_states = encoder_state_model.predict(
    encoder_test[:1],
    verbose=0
)

sample_hidden_state = sample_states[0]
sample_cell_state = sample_states[1]

print("Hidden state shape:",
      sample_hidden_state.shape)

print("Cell state shape:",
      sample_cell_state.shape)

print("""
The encoder processes the input sentence one word at a time.

The encoder produces:

• Hidden state (h): represents the current information.
• Cell state (c): carries longer-term information.

These states are passed to the decoder as context information.
""")


# ============================================================
# CREATE INFERENCE MODELS
# ============================================================

encoder_inference_model = Model(
    encoder_inputs,
    encoder_states
)


decoder_state_h_input = Input(
    shape=(latent_dim,),
    name="Decoder_Hidden_Input"
)

decoder_state_c_input = Input(
    shape=(latent_dim,),
    name="Decoder_Cell_Input"
)

decoder_states_inputs = [
    decoder_state_h_input,
    decoder_state_c_input
]

decoder_single_input = Input(
    shape=(1,),
    name="Decoder_Single_Input"
)

decoder_single_embedding = decoder_embedding_layer(
    decoder_single_input
)

decoder_single_output, decoder_h, decoder_c = decoder_lstm(
    decoder_single_embedding,
    initial_state=decoder_states_inputs
)

decoder_single_output = decoder_dense(
    decoder_single_output
)

decoder_inference_model = Model(
    [
        decoder_single_input,
        decoder_state_h_input,
        decoder_state_c_input
    ],
    [
        decoder_single_output,
        decoder_h,
        decoder_c
    ]
)


# ============================================================
# WORD INDEX DICTIONARIES
# ============================================================

reverse_german_word_index = {
    value: key
    for key, value in german_tokenizer.word_index.items()
}

german_word_index = german_tokenizer.word_index

start_token = german_word_index["<start>"]
end_token = german_word_index["<end>"]


# ============================================================
# TRANSLATION FUNCTION
# ============================================================

def translate_sentence(sentence):

    sentence = clean_text(sentence)

    sequence = english_tokenizer.texts_to_sequences(
        [sentence]
    )

    sequence = pad_sequences(
        sequence,
        maxlen=max_encoder_length,
        padding="post"
    )

    states = encoder_inference_model.predict(
        sequence,
        verbose=0
    )

    target_sequence = np.array(
        [[start_token]]
    )

    translated_words = []

    for _ in range(max_decoder_length):

        output_tokens, h, c = decoder_inference_model.predict(
            [target_sequence, states[0], states[1]],
            verbose=0
        )

        sampled_token_index = np.argmax(
            output_tokens[0, -1, :]
        )

        sampled_word = reverse_german_word_index.get(
            sampled_token_index,
            ""
        )

        if sampled_token_index == end_token:
            break

        if sampled_word not in ["", "<start>"]:
            translated_words.append(
                sampled_word
            )

        target_sequence = np.array(
            [[sampled_token_index]]
        )

        states = [h, c]

    return " ".join(translated_words)


# ============================================================
# Q19 – 10 UNSEEN SENTENCES
# ============================================================

print("\n==========================================")
print("Q19 – UNSEEN SENTENCE TRANSLATIONS")
print("==========================================")

translation_results = []

for i in range(len(english_test)):

    source = english_test[i]

    actual_translation = german_test[i]

    predicted_translation = translate_sentence(
        source
    )

    translation_results.append({
        "English Input": source,
        "Actual German": actual_translation,
        "Predicted German": predicted_translation
    })

translation_df = pd.DataFrame(
    translation_results
)

display(translation_df)


# ============================================================
# Q20 – BLEU SCORE
# ============================================================

print("\n==========================================")
print("Q20 – BLEU SCORE")
print("==========================================")


# Simple BLEU implementation
def ngram_counts(tokens, n):

    counts = {}

    for i in range(len(tokens) - n + 1):

        gram = tuple(
            tokens[i:i+n]
        )

        counts[gram] = counts.get(
            gram,
            0
        ) + 1

    return counts


def sentence_bleu(reference, candidate):

    reference_tokens = reference.lower().split()
    candidate_tokens = candidate.lower().split()

    if len(candidate_tokens) == 0:
        return 0.0

    precisions = []

    for n in range(1, 5):

        reference_counts = ngram_counts(
            reference_tokens,
            n
        )

        candidate_counts = ngram_counts(
            candidate_tokens,
            n
        )

        clipped_count = 0
        total_count = 0

        for gram, count in candidate_counts.items():

            clipped_count += min(
                count,
                reference_counts.get(gram, 0)
            )

            total_count += count

        if total_count == 0:
            precision = 0
        else:
            precision = clipped_count / total_count

        precisions.append(
            precision
        )

    # If any n-gram precision is zero,
    # BLEU becomes zero in this simple implementation.
    if min(precisions) == 0:
        return 0.0

    log_precision = sum(
        np.log(p)
        for p in precisions
    ) / 4

    brevity_penalty = min(
        1,
        np.exp(
            1 - len(reference_tokens) /
            max(len(candidate_tokens), 1)
        )
    )

    bleu = brevity_penalty * np.exp(
        log_precision
    )

    return bleu


bleu_scores = []

for i in range(len(translation_df)):

    reference = translation_df.loc[
        i,
        "Actual German"
    ]

    candidate = translation_df.loc[
        i,
        "Predicted German"
    ]

    score = sentence_bleu(
        reference,
        candidate
    )

    bleu_scores.append(
        score
    )


translation_df["BLEU Score"] = bleu_scores


print("\nTranslation results with BLEU:")
display(
    translation_df.round(4)
)


average_bleu = np.mean(
    bleu_scores
)

print("\nAverage BLEU Score:",
      round(average_bleu, 4))


# ============================================================
# BLEU SCORE GRAPH
# ============================================================

plt.figure(figsize=(12, 5))

plt.bar(
    range(1, len(bleu_scores) + 1),
    bleu_scores
)

plt.title("BLEU Score for 10 Unseen Sentences")
plt.xlabel("Test Sentence")
plt.ylabel("BLEU Score")
plt.grid(axis="y")

plt.show()


# ============================================================
# Q20 – BLEU LIMITATIONS
# ============================================================

print("\n==========================================")
print("BLEU SCORE – LIMITATIONS")
print("==========================================")

print("""
BLEU measures the similarity between a machine-generated
translation and a reference translation using matching
n-grams.

BLEU limitations:

1. It mainly measures word and n-gram overlap.
2. It may give a low score to a correct translation that
   uses different words or sentence structures.
3. It does not fully measure grammatical correctness.
4. It does not completely measure semantic meaning.
5. BLEU depends on the quality and number of reference
   translations.

Therefore, BLEU should be used together with human evaluation
and other translation metrics.
""")


# ============================================================
# Q15–Q20 FINAL SUMMARY
# ============================================================

print("\n==========================================")
print("PART D – FINAL SUMMARY")
print("==========================================")

print("Task: English → German Translation")
print("Architecture: Encoder-Decoder LSTM")
print("Training pairs:", train_size)
print("Unseen test sentences:", len(english_test))
print("Encoder vocabulary:", encoder_vocab_size)
print("Decoder vocabulary:", decoder_vocab_size)
print("Latent dimension:", latent_dim)
print("Training time:", round(training_time, 2), "seconds")
print("Average BLEU:", round(average_bleu, 4))

print("""
Q15 – Sequence-to-sequence translation: COMPLETED
Q16 – Encoder hidden/cell states: COMPLETED
Q17 – Decoder: COMPLETED
Q18 – START/END tokens: COMPLETED
Q19 – 10 unseen sentences: COMPLETED
Q20 – BLEU evaluation and limitations: COMPLETED
""")

print("PART D COMPLETED SUCCESSFULLY!")
```
